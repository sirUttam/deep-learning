## Using nn.Linear instead of managing parameters (weight and bias) manually

```python
#x = torch.tensor([2.0])
target = torch.tensor([4.0])
```

```python
layer = nn.Linear(1,1)
```

```python
optimizer = torch.optim.SGD(layer.parameters(), lr=0.1 ) #layer.parameters() gives optimizer the things it needs -> layer.weight and layer.bias
```

```python
optimizer.zero_grad()
```

```python
y = layer(x)
y
```

```python
loss = (target - y )**2
loss
```

```python
loss.backward()
```

```python
layer.weight
```

```python
layer.weight.grad
```

```python
layer.bias
```

```python
layer.bias.grad
```

```python
loss
```

```python
optimizer.step()
```

```python
optimizer.zero_grad()
```
