```python
w = torch.tensor(0.5, requires_grad=True)
```

```python
w
```

```python
y = w * 2
y
```

```python
target = torch.tensor(4.0)
```

```python
loss = (target - y) ** 2
```

```python
loss
```

```python
loss.backward()    # Calculates gradient
```

```python
w.grad   # Where the gradient is stored
```
