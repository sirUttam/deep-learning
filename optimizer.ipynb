```python
import numpy as np 
import torch
```

```python
w = torch.tensor(0.5, requires_grad=True)
w
```

```python
optimizer = torch.optim.SGD([w], lr=0.1)
```

```python
x = 2
```

## 1st iteration

```python
y = w * x
y
```

```python
target = 4
```

```python
loss = (target - y) ** 2 
loss
```

```python
loss.backward()
```

```python
w.grad
```

```python
optimizer.step()
```

```python
w
```
